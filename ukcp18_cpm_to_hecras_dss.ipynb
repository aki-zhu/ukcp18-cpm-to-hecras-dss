# UKCP18 CPM to HEC-RAS DSS

Convert precipitation with the same configuration and processing code as the command-line interface. Open this notebook from the repository directory.

Edit `config.example.json` to set paths and time windows. Set `output_format` to `dss` or `netcdf`. DSS output is written directly with `pydsstools`; NetCDF export uses the same grid and time settings.

Clipping is optional: set `boundary` to a polygon shapefile, use `choose_boundary: true` to browse, or set `boundary: null` for the full grid. The example uses the broad Forth region from the UK river-basin regions dataset.

Native rotated grids require `target_crs` and `target_cell_size`; see the resolution table in the README.

```python
from pathlib import Path
from ukcp18_cpm_to_hecras_dss import main

CONFIG_PATH = Path("config.example.json").resolve()
RUN = False
INSPECT_ONLY = True  # Set False to convert after checking the metadata.
```

```python
if RUN:
    arguments = ["--config", str(CONFIG_PATH)]
    if INSPECT_ONLY:
        arguments.append("--inspect")
    jobs = main(arguments)
    if not INSPECT_ONLY:
        print(f"Prepared {len(jobs)} job(s).")
else:
    print("Edit the JSON configuration, then set RUN = True.")
```
